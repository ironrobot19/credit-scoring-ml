```python

def create_features(data):
    d = data.copy()

    d["income_monthly"] = d["annual_inc"] / 12
    d["loan_to_income"] = d["loan_amnt"] / (d["annual_inc"] + 1)
    d["payment_to_income"] = d["installment"] / (d["income_monthly"] + 1)
    d["debt_to_income"] = d["dti"] / 100
    d["fico_average"] = (d["fico_range_low"] + d["fico_range_high"]) / 2
    d["fico_spread"] = d["fico_range_high"] - d["fico_range_low"]
    d["credit_per_account"] = d["revol_bal"] / (d["open_acc"] + 1)
    d["account_activity"] = d["open_acc"] / (d["total_acc"] + 1)
    d["delinq_per_account"] = d["delinq_2yrs"] / (d["total_acc"] + 1)
    d["revolving_utilization"] = pd.to_numeric(
        d["revol_util"].astype(str).str.replace("%", ""),
        errors="coerce"
    ) / 100

    return d

train = create_features(train)
valid = create_features(valid)

print("Создано 10 новых финансовых признаков!")
```

```python

!pip -q install optbinning

from optbinning import BinningProcess

new_features = [
    "income_monthly", "loan_to_income",
    "payment_to_income", "debt_to_income",
    "fico_average", "fico_spread",
    "credit_per_account", "account_activity",
    "delinq_per_account", "revolving_utilization"
]

numeric_features = [
    c for c in available_features
    if pd.api.types.is_numeric_dtype(train[c])
]

model_features = list(dict.fromkeys(
    numeric_features + new_features
))

X_train = train[model_features].replace(
    [np.inf, -np.inf], np.nan
)
X_valid = valid[model_features].replace(
    [np.inf, -np.inf], np.nan
)

y_train = train["target"]
y_valid = valid["target"]

binning = BinningProcess(
    variable_names=model_features
)

binning.fit(X_train, y_train)

X_train_woe = binning.transform(
    X_train, metric="woe"
)
X_valid_woe = binning.transform(
    X_valid, metric="woe"
)

print("WoE-биннинг выполнен!")
```

```python

from sklearn.linear_model import LogisticRegression

base_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

base_model.fit(X_train_woe, y_train)

p_valid = base_model.predict_proba(
    X_valid_woe
)[:, 1]

print("Базовая модель обучена!")
```

```python

from sklearn.metrics import roc_auc_score, roc_curve

auc = roc_auc_score(y_valid, p_valid)
gini = 2 * auc - 1

fpr, tpr, thresholds = roc_curve(
    y_valid, p_valid
)
ks = max(tpr - fpr)

print("ROC-AUC:", round(auc, 4))
print("Gini:", round(gini, 4))
print("KS:", round(ks, 4))
```

```python

plt.figure(figsize=(7, 5))

plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
plt.plot([0, 1], [0, 1], "--", color="gray")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC-кривая логистической регрессии")
plt.legend()
plt.grid(True)
plt.show()
```

```python

importance = pd.Series(
    base_model.coef_[0],
    index=model_features
)

importance = importance.abs().sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(9, 5))
importance.sort_values().plot(kind="barh")

plt.title("Топ-10 коэффициентов логистической регрессии")
plt.xlabel("Абсолютная величина коэффициента")
plt.ylabel("Финансовые признаки")
plt.tight_layout()
plt.show()
```

```python

!pip -q install lightgbm

import lightgbm as lgb

boost_model = lgb.LGBMClassifier(
    n_estimators=1000,
    learning_rate=0.03,
    num_leaves=31,
    random_state=42,
    verbosity=-1
)

boost_model.fit(
    X_train_woe, y_train,
    eval_set=[(X_valid_woe, y_valid)],
    eval_metric="auc",
    callbacks=[lgb.early_stopping(50, verbose=False)]
)

p_boost = boost_model.predict_proba(
    X_valid_woe
)[:, 1]

print("LightGBM успешно обучен!")
```

```python

from sklearn.metrics import roc_auc_score, roc_curve

def calculate_metrics(y, p):
    auc = roc_auc_score(y, p)
    fpr, tpr, _ = roc_curve(y, p)
    return [auc, 2 * auc - 1, max(tpr - fpr)]

results = pd.DataFrame(
    [
        calculate_metrics(y_valid, p_valid),
        calculate_metrics(y_valid, p_boost)
    ],
    columns=["ROC-AUC", "Gini", "KS"],
    index=["Logistic Regression", "LightGBM"]
)

display(results.round(4))
```

```python

def calculate_psi(expected, actual, bins=10):
    expected = np.asarray(expected, dtype=float)
    actual = np.asarray(actual, dtype=float)

    expected = expected[np.isfinite(expected)]
    actual = actual[np.isfinite(actual)]

    cuts = np.unique(
        np.quantile(expected, np.linspace(0, 1, bins+1))
    )
    cuts[0], cuts[-1] = -np.inf, np.inf

    e = np.histogram(expected, cuts)[0] / len(expected)
    a = np.histogram(actual, cuts)[0] / len(actual)

    e = np.clip(e, 1e-6, None)
    a = np.clip(a, 1e-6, None)

    return np.sum((a-e) * np.log(a/e))

p_train = boost_model.predict_proba(
    X_train_woe
)[:, 1]

print("PSI прогнозов:",
      round(calculate_psi(p_train, p_test), 4))

importance = pd.Series(
    boost_model.feature_importances_,
    index=model_features
).nlargest(5)

for col in importance.index:
    value = calculate_psi(
        X_train_woe[col],
        X_test_woe[col]
    )
    print(col, "PSI:", round(value, 4))
```

```python

thresholds = np.arange(0.05, 0.96, 0.05)
profits = []

loan_amount = test["loan_amnt"].to_numpy()
actual_default = y_test.to_numpy()

for threshold in thresholds:
    approved = p_test < threshold

    profit = np.where(
        actual_default == 0,
        loan_amount * 0.15,
        -loan_amount * 0.45
    )

    profits.append(profit[approved].sum())

best_idx = np.argmax(profits)

print("Лучший порог:", round(thresholds[best_idx], 2))
print("Максимальная прибыль:",
      round(profits[best_idx], 2), "USD")

plt.figure(figsize=(8, 5))
plt.plot(thresholds, profits, marker="o")
plt.axvline(thresholds[best_idx],
            color="red", linestyle="--")
plt.xlabel("Порог вероятности дефолта")
plt.ylabel("Прибыль, USD")
plt.title("Прибыль банка при разных порогах")
plt.grid(True)
plt.show()
```

```python

thresholds = np.arange(0.05, 0.96, 0.05)

def bank_profit(y, p, amounts, threshold):
    approved = p < threshold
    profit = np.where(
        np.asarray(y) == 0,
        np.asarray(amounts) * 0.15,
        -np.asarray(amounts) * 0.45
    )
    return profit[approved].sum()

profits_valid = [
    bank_profit(
        y_valid, p_boost,
        valid["loan_amnt"], t
    )
    for t in thresholds
]

best_threshold = thresholds[np.argmax(profits_valid)]

print("Оптимальный порог на VALIDATION:",
      round(best_threshold, 2))

print("Прибыль на VALIDATION:",
      round(max(profits_valid), 2), "USD")
```

```python

final_profit = bank_profit(
    y_test, p_test,
    test["loan_amnt"],
    best_threshold
)

approved = p_test < best_threshold

print("Выбранный порог:", round(best_threshold, 2))
print("Прибыль на TEST:", round(final_profit, 2), "USD")
print("Одобрено кредитов:", approved.sum())
print("Доля одобрения:",
      round(approved.mean() * 100, 2), "%")
```
